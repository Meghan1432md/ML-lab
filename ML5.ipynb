```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (12.0, 9.0)
np.random.seed(69)
X = np.random.rand(100) * 10
Y = 2 * X + 1 + np.random.randn(100) * 2
plt.scatter(X, Y)
plt.show()

# Building the model
w1_old = 0.1
w0_old= 0.1
w1_new = 0
w0_new= 0
alpha = 0.01  # The learning Rate
epochs = 1000  # The number of iterations to perform gradient descent
n = float(len(X)) # Number of elements in X
# Performing Gradient Descent
while(abs(w1_new - w1_old) + abs(w0_new - w0_old))>0.01:
  w1_old=w1_new
  w0_old = w0_new
  Y_pred = w1_old*X + w0_old  # The current predicted value of Y
  D_w1 = (-2/n) * sum(X * (Y - Y_pred))  # Derivative wrt w1
  D_w0 = (-2/n) * sum(Y - Y_pred)  # Derivative wrt w0
  w1_new = w1_old - alpha * D_w1  # Update w1
  w0_new= w0_old - alpha * D_w0  # Update c
print ("Final coefficients: w1 =", w1_new, ", w0 =", w0_new)
# Making predictions
Y_pred = w1_new*X + w0_new # Use the updated w1_new and w0_new for prediction
plt.scatter(X, Y)
plt.plot([min(X), max(X)], [min(Y_pred), max(Y_pred)],
color='red')  # regression line
plt.show()
```
