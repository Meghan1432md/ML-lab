```python
 import numpy as np
 import matplotlib.pyplot as plt
 from sklearn import datasets
 from sklearn.svm import SVC
 from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier
 from sklearn.preprocessing import StandardScaler
 # 1. Load and Scale Data
 iris = datasets.load_iris()
 X = iris.data[:, [2, 3]] # Petal Length and Width
 y = iris.target
 scaler = StandardScaler()
 X_scaled = scaler.fit_transform(X)
 # 2. Define the two Multiclass Strategies
 base_svm = SVC(kernel=’linear’, C=1.0)
 ovr_model = OneVsRestClassifier(base_svm).fit(X_scaled, y)
 ovo_model = OneVsOneClassifier(base_svm).fit(X_scaled, y)
 def plot_multiclass_comparison(models, titles, X, y, target_names):
fig, axes = plt.subplots(1, 2, figsize=(16, 7))
colors = [’purple’, ’teal’, ’orange’]
h = 0.02
x_min, x_max = X[:, 0].min()- 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min()- 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min,
y_max, h))
grid_points = np.c_[xx.ravel(), yy.ravel()]
for ax, model, title in zip(axes, models, titles):
Z_class = model.predict(grid_points).reshape(xx.shape)
ax.contourf(xx, yy, Z_class, cmap=plt.cm.coolwarm, alpha=0.1)
for i, name in enumerate(target_names):
ax.scatter(X[y == i, 0], X[y == i, 1], label=name,
color=colors[i], edgecolors=’k’, zorder=5)
for i, estimator in enumerate(model.estimators_):
Z = estimator.decision_function(grid_points).reshape(xx.
shape)
ax.contour(xx, yy, Z, levels=[0], colors=’black’,
linewidths=1.5, zorder=4)
line_color = colors[i % 3]
ax.contour(xx, yy, Z, levels=[-1, 1], colors=line_color,
linestyles=’:’, linewidths=1.2, alpha=0.7)
ax.set_title(title, fontsize=14, fontweight=’bold’)
ax.set_xlabel("Petal Length (Standardized)")
ax.set_ylabel("Petal Width (Standardized)")
ax.legend(loc=’upper left’)
plt.tight_layout()
plt.show()
plot_multiclass_comparison([ovr_model, ovo_model],
["One-vs-Rest (OvR) Strategy", "One-vs-One (OvO) Strategy"],
X_scaled, y, iris.target_names)
```
